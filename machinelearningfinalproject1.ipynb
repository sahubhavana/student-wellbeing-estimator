# DATA CLEANING

```python
df.drop_duplicates(inplace=True)
```

```python
df.describe()
```

```python
df['Physical_Activity_Hours']=df['Physical_Activity_Hours'].clip(lower=0)
```

```python
df.describe()
```

```python
top_counteries=df['Country'].value_counts().index[:10].tolist()
top_counteries
```

```python
def group_countries(Country):
    if Country in top_counteries:
        return Country
    else:
        return 'Other'
        
```

```python
df['group_country']=df['Country'].apply(group_countries)
```

```python
df['group_country'].value_counts()
```

# train test split

```python
skweecol=['Study_Hours']
other_numeric_col=["Age",'Avg_Daily_Usage_Hours', 'Daily_Unlocks'
       , 'Physical_Activity_Hours', 'Sleep_Hours_Per_Night']
ordinal_col=['Stress_Level']
normal_col=['Gender','Academic_Level','Most_Used_Platform','Purpose_Of_Use','group_country']

feature_col=skweecol+other_numeric_col+ordinal_col+normal_col
x=df[feature_col]
y=df['Mental_Health_Score']
```

# pipeline1: for skeweed data
# pipeline2: for numeric data
# pipeline3: ordinal data
# pipeline4: nominal data

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer,StandardScaler
from sklearn.preprocessing import OrdinalEncoder,OneHotEncoder
from sklearn.compose import ColumnTransformer
skew_pipeline=Pipeline(steps=[('log_transform',FunctionTransformer(np.log1p)),('scale',StandardScaler())])
plain_numeric_pipeline=Pipeline(steps=[('scale',StandardScaler())])
ordinal_pipeline=Pipeline(steps=[('encode',OrdinalEncoder(categories=[['Low','Medium','High','Very High']]))])
nominal_pipeline = Pipeline(steps=[
    ('encode', OneHotEncoder(handle_unknown='ignore'))
])


preprocessor=ColumnTransformer(transformers=[("skewed_pipeline",skew_pipeline,skweecol),
                                ('plain_numeric',plain_numeric_pipeline,other_numeric_col),
                                ('ordinal',ordinal_pipeline,ordinal_col),
                                ('normal',nominal_pipeline,normal_col)
        ])
preprocessor
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
lr_pipeline=Pipeline(steps=[('preprocessor',preprocessor),
                            ('linearreggerssor',LinearRegression())])
```

```python
lr_pipeline.fit(x_train,y_train)
```

```python
lr_pipeline.predict(x_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
lr_preds=lr_pipeline.predict(x_test)
lr_preds_train=lr_pipeline.predict(x_train)

lr_l2_test=r2_score(y_test,lr_preds)
lr_l2_trainning=r2_score(y_train,lr_preds_train)


print(f"Accuracy of trainning=",lr_l2_test)
print(f"Accuracy of testing=",lr_l2_trainning)
```

```python
from sklearn.ensemble import RandomForestRegressor
rf_pipeline=Pipeline(steps=[('preprocessor',preprocessor),
                            ('regressor',RandomForestRegressor())])
```

```python
rf_pipeline.fit(x_train,y_train)
```

```python
lr_preds=rf_pipeline.predict(x_test)
lr_preds_train=rf_pipeline.predict(x_train)

lr_l2_test=r2_score(y_test,lr_preds)
lr_l2_trainning=r2_score(y_train,lr_preds_train)


print(f"Accuracy of trainning=",lr_l2_test)
print(f"Accuracy of testing=",lr_l2_trainning)
```

```python
from sklearn.model_selection import RandomizedSearchCV
param_grid = {
    'regressor__n_estimators': [100, 200, 300],
    'regressor__max_depth': [5, 10, 15],
    'regressor__min_samples_split': [2, 5, 10],
    'regressor__min_samples_leaf': [1, 2, 4]
}

rm = RandomizedSearchCV(
    estimator=rf_pipeline,
    param_distributions=param_grid,
    n_iter=15,
    cv=5,
    scoring='r2',
    random_state=42,
    n_jobs=-1
)

rm.fit(x_train, y_train)

print("Best Parameters:", rm.best_params_)
print("Best R2 Score:", rm.best_score_)

rm.fit(x_train,y_train)
```
